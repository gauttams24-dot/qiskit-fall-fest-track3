```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.circuit.library import PauliEvolutionGate
from qiskit.synthesis import LieTrotter
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime.fake_provider import FakeSherbrooke
from qiskit_ibm_runtime import EstimatorV2 as Estimator

# 1. MAP: 4-Spin 1D Transverse-Field Ising Chain: H = -J sum(Z_i Z_{i+1}) - h sum(X_i)
num_spins = 4
J, h = 1.0, 0.5
zz_terms = [("ZZ", [i, i + 1], -J) for i in range(num_spins - 1)]
x_terms = [("X", [i], -h) for i in range(num_spins)]
hamiltonian = SparsePauliOp.from_sparse_list(zz_terms + x_terms, num_qubits=num_spins)

# Observable: Average Magnetization along Z
magnetization = SparsePauliOp.from_sparse_list([("Z", [i], 1.0 / num_spins) for i in range(num_spins)], num_qubits=num_spins)

# Build Trotterized circuit for time t = 1.0 with 4 Trotter steps
t = 1.0
trotter_steps = 4
qc = QuantumCircuit(num_spins)
evo_gate = PauliEvolutionGate(hamiltonian, time=t, synthesis=LieTrotter(reps=trotter_steps))
qc.append(evo_gate, range(num_spins))

# 2. OPTIMIZE (Transpile for noisy hardware backend)
backend = FakeSherbrooke()
pm = generate_preset_pass_manager(backend=backend, optimization_level=3)
isa_circuit = pm.run(qc)
isa_observable = magnetization.apply_layout(isa_circuit.layout)

# 3. EXECUTE on Noisy Backend
estimator = Estimator(mode=backend)
job = estimator.run([(isa_circuit, isa_observable)])
result = job.result()[0]
print(f"Noisy Expectation Value <Mz> at t={t}: {result.data.evs:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.circuit.library import PauliEvolutionGate
from qiskit.synthesis import LieTrotter
from qiskit.primitives import StatevectorEstimator
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime.fake_provider import FakeSherbrooke
from qiskit_ibm_runtime import EstimatorV2 as Estimator

# 1. DEFINE 1D ISING SPIN CHAIN SIMULATION (N=4 Spins)
num_spins = 4
J, h = 1.0, 0.5
zz_terms = [("ZZ", [i, i + 1], -J) for i in range(num_spins - 1)]
x_terms = [("X", [i], -h) for i in range(num_spins)]
hamiltonian = SparsePauliOp.from_sparse_list(zz_terms + x_terms, num_qubits=num_spins)
magnetization = SparsePauliOp.from_sparse_list([("Z", [i], 1.0 / num_spins) for i in range(num_spins)], num_qubits=num_spins)

# 2. RUN ACROSS TIME STEPS WITH REAL STATEVECTOR & REAL ZNE GATE FOLDING
time_steps = np.linspace(0.0, 2.0, 6)
ideal_magnetization = []
noisy_magnetization = []
mitigated_magnetization = []

backend = FakeSherbrooke()
pm = generate_preset_pass_manager(backend=backend, optimization_level=1)
noisy_estimator = Estimator(mode=backend)
exact_estimator = StatevectorEstimator()

for t in time_steps:
    if t == 0.0:
        ideal_magnetization.append(1.0)
        noisy_magnetization.append(1.0)
        mitigated_magnetization.append(1.0)
        continue

    # Base Trotterized circuit (Noise Scale lambda = 1)
    qc = QuantumCircuit(num_spins)
    evo_gate = PauliEvolutionGate(hamiltonian, time=t, synthesis=LieTrotter(reps=3))
    qc.append(evo_gate, range(num_spins))

    # A. Exact Noiseless Quantum Statevector Simulation
    ideal_val = float(exact_estimator.run([(qc, magnetization)]).result()[0].data.evs)
    ideal_magnetization.append(ideal_val)

    # B. Noisy Hardware Simulation (lambda = 1) on FakeSherbrooke
    isa_qc_1 = pm.run(qc)
    isa_obs_1 = magnetization.apply_layout(isa_qc_1.layout)
    val_lambda_1 = float(noisy_estimator.run([(isa_qc_1, isa_obs_1)]).result()[0].data.evs)
    noisy_magnetization.append(val_lambda_1)

    # C. Real ZNE Unitary Gate Folding (U -> U * U^dagger * U for lambda = 3)
    qc_folded = QuantumCircuit(num_spins)
    qc_folded.append(evo_gate, range(num_spins))
    qc_folded.barrier()
    qc_folded.append(evo_gate.inverse(), range(num_spins))
    qc_folded.barrier()
    qc_folded.append(evo_gate, range(num_spins))

    isa_qc_3 = pm.run(qc_folded)
    isa_obs_3 = magnetization.apply_layout(isa_qc_3.layout)
    val_lambda_3 = float(noisy_estimator.run([(isa_qc_3, isa_obs_3)]).result()[0].data.evs)

    # Richardson Linear Extrapolation to Zero Noise (lambda -> 0)
    zne_val = (3.0 * val_lambda_1 - val_lambda_3) / 2.0
    mitigated_magnetization.append(zne_val)

# 3. PLOT RESULTS FOR JUDGES
plt.figure(figsize=(9, 5))
plt.plot(time_steps, ideal_magnetization, 'k--', label='Exact Statevector (Noiseless)', linewidth=2)
plt.plot(time_steps, noisy_magnetization, 'r-o', label='Noisy Hardware (FakeSherbrooke, λ=1)', linewidth=2)
plt.plot(time_steps, mitigated_magnetization, 'g-s', label='ZNE Error-Mitigated (Gate Folding λ=1,3 → 0)', linewidth=2)
plt.title('Track 3: 1D Ising Spin Chain Dynamics & ZNE Error Mitigation', fontsize=12)
plt.xlabel('Time (t)', fontsize=11)
plt.ylabel('Average Magnetization <Mz>', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=10)
plt.show()
```
